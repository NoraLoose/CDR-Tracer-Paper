# Computational savings analysis (ROMS/MARBL)

Raw timing data extracted from `MPI_run_time` / `ntimes` in the `output.log` files under `roms_marbl_dic/exp*` (DOR-forcing baseline), `roms_marbl_alk/exp*` (OAE-forcing baseline), and `roms/exp*` (DOR- and OAE-forced runs with 2x12 embedded mCDR response tracers, `ANTITRACER` config).

Only runs with a completed `output.log` (i.e. a recorded `MPI_run_time`) are listed. Directories with no output at all (run never completed, or logs since removed) are omitted entirely rather than tracked as rows.

All runs used 7 nodes x 128 cores = 896 cores on Perlmutter, dt = 900 s.

`simulated_days = ntimes * dt_sec / 86400` is computed exactly per run (it varies a little in principle, e.g. a run that stops early), and

`core_hours_per_sim_year = cores * mpi_run_time_sec * (365 / simulated_days) / 3600`

rather than assuming every baseline run is exactly 2 months and every tracer run exactly 24 months.

```python
import pandas as pd

CORES = 7 * 128
DAYS_PER_YEAR = 365.0

raw = [
    # group, site, log_path, machine, ntimes, dt_sec, mpi_run_time_sec, included, exclusion_reason
    ("baseline_dor", "BC10", "roms_marbl_dic/expBC10/OUTPUT/output.log", "perlmutter", 5856, 900, 18907.710334290001, True, None),
    ("baseline_dor", "BC7",  "roms_marbl_dic/expBC7/OUTPUT/output.log",  "perlmutter", 5856, 900, 18407.189070027001, True, None),
    ("baseline_dor", "BC8",  "roms_marbl_dic/expBC8/OUTPUT/output.log",  "perlmutter", 5856, 900, 18495.149602092002, True, None),
    ("baseline_dor", "EC10", "roms_marbl_dic/expEC10/OUTPUT/output.log", "perlmutter", 5856, 900, 18945.762408160001, True, None),
    ("baseline_dor", "EC7",  "roms_marbl_dic/expEC7/OUTPUT/output.log",  "perlmutter", 5856, 900, 18652.749636027002, True, None),
    ("baseline_dor", "EC8",  "roms_marbl_dic/expEC8/OUTPUT/output.log",  "perlmutter", 5856, 900, 18663.107204344000, True, None),
    ("baseline_dor", "JP10", "roms_marbl_dic/expJP10/OUTPUT/output.log", "perlmutter", 5856, 900, 18443.350212056001, True, None),
    ("baseline_dor", "JP7",  "roms_marbl_dic/expJP7/OUTPUT/output.log",  "perlmutter", 5856, 900, 18495.567425382000, True, None),
    ("baseline_dor", "JP8",  "roms_marbl_dic/expJP8/OUTPUT/output.log",  "perlmutter", 5856, 900, 18424.221101873998, True, None),
    ("baseline_dor", "VI10", "roms_marbl_dic/expVI10/OUTPUT/output.log", "perlmutter", 5856, 900, 18464.383533058000, True, None),
    ("baseline_dor", "VI7",  "roms_marbl_dic/expVI7/OUTPUT/output.log",  "perlmutter", 5856, 900, 18490.079755885003, True, None),
    ("baseline_dor", "VI8",  "roms_marbl_dic/expVI8/OUTPUT/output.log",  "perlmutter", 5856, 900, 18473.145958665002, True, None),

    ("baseline_oae", "BC8", "roms_marbl_alk/expBC8_perlmutter/OUTPUT/output.log", "perlmutter", 5856, 900, 18433.289086269000, True, None),
    ("baseline_oae", "EC8", "roms_marbl_alk/expEC8_perlmutter/OUTPUT/output.log", "perlmutter", 5856, 900, 18529.492123329997, True, None),
    ("baseline_oae", "JP8", "roms_marbl_alk/expJP8_perlmutter/OUTPUT/output.log", "perlmutter", 5856, 900, 19099.852160424998, True, None),
    ("baseline_oae", "VI8", "roms_marbl_alk/expVI8_perlmutter/OUTPUT/output.log", "perlmutter", 5856, 900, 18503.861234453001, True, None),

    ("cdr_tracer_dor", "exp0", "roms/exp0/OUTPUT_dor/output.log", "perlmutter", 70368, 900, 131931.84242529800, True, None),
    ("cdr_tracer_dor", "exp1", "roms/exp1/OUTPUT_dor/output.log", "perlmutter", 70368, 900, 131882.78473718200, True, None),
    ("cdr_tracer_dor", "exp2", "roms/exp2/OUTPUT_dor/output.log", "perlmutter", 70368, 900, 131674.65698570499, True, None),
    ("cdr_tracer_dor", "exp6", "roms/exp6/OUTPUT_dor/output.log", "perlmutter", 70368, 900, 131651.77466204800, True, None),

    ("cdr_tracer_oae", "exp2", "roms/exp2/OUTPUT_oae/output.log", "perlmutter", 70368, 900, 150644.64924403001, True, None),
    ("cdr_tracer_oae", "exp3", "roms/exp3/OUTPUT_oae/output.log", "perlmutter", 70368, 900, 151116.97527306300, True, None),
]

df = pd.DataFrame(raw, columns=[
    "group", "site", "log_path", "machine", "ntimes", "dt_sec",
    "mpi_run_time_sec", "included", "exclusion_reason",
])
df["simulated_days"] = df["ntimes"] * df["dt_sec"] / 86400
df
```

## Per-run core-hours/sim-year

Uses each run's own `simulated_days` rather than an assumed 2-month / 24-month duration, so it's robust to runs of slightly different length.

```python
df["core_hours_per_sim_year"] = (
    CORES * df["mpi_run_time_sec"] * (DAYS_PER_YEAR / df["simulated_days"]) / 3600
)
df
```

## Group summary and final table

Baseline (DOR + OAE forcing) is pooled into a single average, since forcing type barely affects baseline cost.
DOR- and OAE-forced tracer runs are kept as separate rows, since the ~14% cost gap between them is systematic, not noise.

```python
included = df[df["included"]]

baseline = included[included["group"].isin(["baseline_dor", "baseline_oae"])]
cdr_tracer_dor = included[included["group"] == "cdr_tracer_dor"]
cdr_tracer_oae = included[included["group"] == "cdr_tracer_oae"]

N_BASELINE_EXPERIMENTS = 12  # one per site, matches roms_marbl_dic site count

summary_rows = []

ch_yr = baseline["core_hours_per_sim_year"].mean()
summary_rows.append({
    "configuration": "ROMS/MARBL (baseline)",
    "n_samples": len(baseline),
    "core_hours_per_sim_year": ch_yr,
    "n_experiments": N_BASELINE_EXPERIMENTS,
    "total_core_hours": ch_yr * N_BASELINE_EXPERIMENTS,
    "speedup_vs_baseline": None,
})
baseline_total = summary_rows[0]["total_core_hours"]

for label, sub in [
    ("ROMS DOR + 2x12 mCDR response tracers", cdr_tracer_dor),
    ("ROMS OAE + 2x12 mCDR response tracers", cdr_tracer_oae),
]:
    ch_yr = sub["core_hours_per_sim_year"].mean()
    total = ch_yr * 1
    summary_rows.append({
        "configuration": label,
        "n_samples": len(sub),
        "core_hours_per_sim_year": ch_yr,
        "n_experiments": 1,
        "total_core_hours": total,
        "speedup_vs_baseline": baseline_total / total,
    })

summary = pd.DataFrame(summary_rows).round(1)
summary
```

## CESM group summary and final table

Baseline is a single measured 16-year run (`control.001_backup`), not your colleague's original 483 estimate. DOR and OAE tracer segments are each averaged over their 15 yearly job chunks and kept as separate rows.

```python
cesm_baseline = cesm_df[cesm_df["group"] == "baseline"]
cesm_tracer_dor = cesm_df[cesm_df["group"] == "cdr_tracer_dor"]
cesm_tracer_oae = cesm_df[cesm_df["group"] == "cdr_tracer_oae"]

N_CESM_BASELINE_EXPERIMENTS = 690  # one per response tracer

cesm_summary_rows = []

ch_yr = cesm_baseline["core_hours_per_sim_year"].mean()
cesm_summary_rows.append({
    "configuration": "CESM/MARBL (baseline)",
    "n_samples": len(cesm_baseline),
    "core_hours_per_sim_year": ch_yr,
    "n_experiments": N_CESM_BASELINE_EXPERIMENTS,
    "total_core_hours": ch_yr * N_CESM_BASELINE_EXPERIMENTS,
    "speedup_vs_baseline": None,
})
cesm_baseline_total = cesm_summary_rows[0]["total_core_hours"]

for label, sub in [
    ("CESM DOR + 1x690 mCDR response tracers", cesm_tracer_dor),
    ("CESM OAE + 1x690 mCDR response tracers", cesm_tracer_oae),
]:
    ch_yr = sub["core_hours_per_sim_year"].mean()
    total = ch_yr * 1
    cesm_summary_rows.append({
        "configuration": label,
        "n_samples": len(sub),
        "core_hours_per_sim_year": ch_yr,
        "n_experiments": 1,
        "total_core_hours": total,
        "speedup_vs_baseline": cesm_baseline_total / total,
    })

cesm_summary = pd.DataFrame(cesm_summary_rows).round(1)
cesm_summary
```
